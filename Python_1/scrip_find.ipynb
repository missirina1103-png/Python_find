```python
s='I love coffee and coffee makes my mornings better. My favorite coffee is raf'

ind_1 = s.find('coffee')
print("1st coffee", ind_1)

ind_2 = s.rfind('coffee')
print('Last coffee', ind_2)

ind_3 = s.find('latte')
print('Latte', ind_3)

ind_4=s.find('coffee', 10)
print("2st coffee", ind_4)

ind_5=s.find('coffee', 10, 60)
print("2st coffee", ind_5)

ind_6 = s.find('.')
subtext_1 = s[:ind_6]
print('Substring:', subtext_1)

subtext_2= s[0::2]
print('Every second character:', subtext_2)

start = s.find('and')
subtext_3 = s[start:]
print("From 'and':", subtext_3)
```
