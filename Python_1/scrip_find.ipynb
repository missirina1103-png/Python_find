```python
s = "I love coffee and coffee makes my mornings better. My favorite coffee is raf"

print("=" * 60)
print("ORIGINAL STRING")
print("=" * 60)
print(s)
print("Length:", len(s), "characters\n")

ind_1 = s.find("coffee")
ind_2 = s.rfind("coffee")
print("First 'coffee' at index:", ind_1)
print("Last  'coffee' at index:", ind_2)

ind_3 = s.find("latte")
print("\nSearch for 'latte' :", ind_3, "(means: not found)")

ind_4 = s.find("coffee", 10)
print("\n'coffee' starting from index 10 :", ind_4)

ind_5 = s.find("coffee", 10, 60)
print("'coffee' inside [10, 60)       :", ind_5)

print("\nAll 'coffee' positions:")
start = 0
positions = []
while True:
    pos = s.find("coffee", start)
    if pos == -1:
        break
    positions.append(pos)
    start = pos + 1

for p in positions:
    print(f"  index {p:>3}  ->  ...{s[max(0, p-3):p+9]}...")

print("Total occurrences:", len(positions))

first_dot = s.find(".")
print("\nFirst '.' at index:", first_dot)

subtext_1 = s[:first_dot]
print("Substring before first '.':", subtext_1)

subtext_2 = s[0::2]
print("Every 2nd character       :", subtext_2)

start_and = s.find("and")
subtext_3 = s[start_and:]
print("From 'and' to the end     :", subtext_3)

last_coffee = s.rfind("coffee")
after_last = s[last_coffee + len("coffee"):]
print("\nAfter last 'coffee':", after_last.strip())

word = "coffee"
count = 0
i = 0
while i < len(s):
    j = s.find(word, i)
    if j == -1:
        break
    count += 1
    i = j + 1
print(f"\nOccurrences of '{word}':", count)

lower_s = s.lower()
print("\nCase-insensitive search for 'COFFEE' at:", lower_s.find("coffee"))

print("\n" + "=" * 60)
user_word = input("Enter a word to search in the string: ").strip()
if user_word:
    idx = s.find(user_word)
    if idx == -1:
        print(f"'{user_word}' was not found.")
    else:
        print(f"First '{user_word}' at {idx}")
        print(f"Last  '{user_word}' at {s.rfind(user_word)}")
```
